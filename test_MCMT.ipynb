```python
from __Detection.detection_api import DetectionAPI
from __Tracking.tracking_api import TrackerAPI
from tools.homo_graphy import PlanProjector
import cv2
```

```python
class Args:
    track_thresh = 0.5
    match_thresh = 0.5
    track_buffer = 60
    mot20 = False
    cpu_workers = 10   # 듀얼 CPU 적극 활용
    chunk_sec   = 10.0 # 15~30초 권장
args = Args()


CLASSES= ["People"]
```

```python
detector = DetectionAPI(model=["ultra_people", "worker"], class_names=CLASSES)
tracker = TrackerAPI(args=args, detector=detector)
```

```python
# 1) 도면 + (선택) H 로드
plan_path = "/workspace/assets/250904_homograph_coordinate-plane2.jpg"
plan_pts  = [
                (1170,  214),    #point1
                (1170,  559),   #point2
                (1170,  904),   #point3
                (2212,  214),    #point4
                (2212,  559),   #point5
                (2212,  904),   #point6
                (3255,  214),     #point7
                (3255,  559),    #point8
                (3255,  904),    #point9
                ]
```

### Cam#1

```python
cam1_tracklets = tracker.track_video(
    video_path="/workspace/datasets/homography_experiment2/1100-1110/2025-09-04 10_59_59 이동형 #1_part_000.mp4",
    save_path="/workspace/results/tracking_result1.mp4",
    trail_len=100
)
```

```python
# 2) 대응점으로 H 추정 (리스트 안에 '튜플' 형태) --> 예시 나중에 바꿔야함
projector1 = PlanProjector(plan_path, trail_len=60, trail_ttl=30, line_thickness=4, point_radius=6)
cam1_pts = [(1033,475),
            (948,474),
            (863,473),
            (1019,527),
            (890,524),
            (769,519),
            (973,667),
            (741,652),
            (548,620)]
H1, mask1 = projector1.fit_homography(cam1_pts, plan_pts, ransac_thresh=3.0)

# 4-1) 예시, 한 프레임만 투영(10번 프레임) & 저장
projected1, canvas1 = projector1.projection(cam1_tracklets[10], mode="bottom-center", draw=True)
cv2.imwrite("/workspace/results/plan_proj_frame1.jpg", canvas1)
# 4-2) 전체 프레임 미니맵 비디오 저장
projector1.save_video(cam1_tracklets, "/workspace/results/plan_projection1.mp4", fps=30, mode="bottom-center")
```
