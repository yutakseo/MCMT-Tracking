```python
import subprocess
import shutil
from pathlib import Path

def split_video(input_file: str, chunk_minutes: int = 10, reencode: bool = False):
    """
    영상 파일을 일정 분 단위로 자르는 함수.

    Parameters
    ----------
    input_file : str
        자를 영상 파일 경로
    chunk_minutes : int
        분할 단위 (분). 기본: 10분
    reencode : bool
        True이면 재인코딩해서 정확히 자름 (느림, 품질변화 가능)
        False이면 키프레임 기준 무손실 컷 (빠름, 경계가 살짝 어긋날 수 있음)
    """
    if shutil.which("ffmpeg") is None:
        raise RuntimeError("ffmpeg가 설치되어 있지 않습니다. (https://ffmpeg.org)")

    input_path = Path(input_file).resolve()
    out_dir = input_path.parent / f"{input_path.stem}_chunks"
    out_dir.mkdir(exist_ok=True)

    chunk_seconds = chunk_minutes * 60
    output_pattern = str(out_dir / f"{input_path.stem}_part_%03d{input_path.suffix}")

    cmd = ["ffmpeg", "-hide_banner", "-y", "-i", str(input_path), "-map", "0"]

    if not reencode:
        # 무손실 빠른 컷
        cmd += [
            "-c", "copy",
            "-f", "segment",
            "-segment_time", str(chunk_seconds),
            "-reset_timestamps", "1",
            output_pattern
        ]
    else:
        # 정확히 10분 단위 컷 (재인코딩)
        cmd += [
            "-c:v", "libx264", "-preset", "veryfast", "-crf", "23",
            "-c:a", "aac", "-b:a", "192k",
            "-force_key_frames", f"expr:gte(t,n_forced*{chunk_seconds})",
            "-f", "segment",
            "-segment_time", str(chunk_seconds),
            "-reset_timestamps", "1",
            output_pattern
        ]

    print(">> 실행:", " ".join(cmd))
    subprocess.run(cmd, check=True)
    print(f"완료! 결과는 {out_dir} 에 저장됨")


```
